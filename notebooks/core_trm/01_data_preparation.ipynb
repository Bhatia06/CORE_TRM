# CORE-trm: Data Preparation

Same data preparation as CORE-ave. This notebook can be skipped if CORE-ave data preparation was already run.

````python
# This notebook uses the same data preparation as CORE-ave.
# If you already ran the CORE-ave data preparation notebook,
# the processed data is already available.

import pandas as pd
import json
from pathlib import Path

PROCESSED_DIR = Path("../../data/processed")

# Verify processed data exists
for f in ["train.parquet", "validation.parquet", "test.parquet", "item_mapping.json"]:
    path = PROCESSED_DIR / f
    exists = path.exists()
    print(f"{f}: {'EXISTS' if exists else 'MISSING'}")

if exists:
    with open(PROCESSED_DIR / "item_mapping.json") as f:
        mapping = json.load(f)
    print(f"\nVocab size: {mapping['vocab_size']}")
    print(f"Min train freq: {mapping['min_train_freq']}")
    
    train = pd.read_parquet(PROCESSED_DIR / "train.parquet")
    val = pd.read_parquet(PROCESSED_DIR / "validation.parquet")
    test = pd.read_parquet(PROCESSED_DIR / "test.parquet")
    print(f"\nTrain: {len(train):,} pairs")
    print(f"Val: {len(val):,} pairs")
    print(f"Test: {len(test):,} pairs")
````
